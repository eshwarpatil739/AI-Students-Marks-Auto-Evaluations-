```python
nlp = spacy.load("en_core_web_sm")

def preprocess(text):
    doc = nlp(text.lower())
    tokens = [
        token.lemma_
        for token in doc
        if token.is_alpha and not token.is_stop
    ]
    return " ".join(tokens)
```

```python
from sentence_transformers import SentenceTransformer , util
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
model
```

```python
def compute_semantic_score(row):
    student_answer = str(row["Student Answer"]).strip()
    model_answer = str(row["Model Answer"]).strip()

    if student_answer == "" or model_answer == "":
        return 0.0

    embeddings = model.encode(
        [student_answer, model_answer],
        convert_to_tensor=True
    )

    similarity = util.cos_sim(embeddings[0], embeddings[1]).item()

    semantic_max = float(row["max_marks"]) * 0.80
    semantic_score = similarity * semantic_max

    return round(semantic_score, 2)
```

```python
df3["semantic_score"] = df3.apply(compute_semantic_score, axis=1)
df3["semantic_score"].head()
```

```python
def compute_structure_score(row):
    try:
        answer = str(row["Student Answer"]).strip()
        word_count = len(answer.split())
        max_marks_structure = float(row["max_marks"]) * 0.10

        ratio = min(word_count / 30 , 1)
        return round(max_marks_structure * ratio , 2)

    except Exception:
        return 0.0
```

```python
df3["answer_structure_score"] = df3.apply(compute_structure_score, axis = 1)
df3["answer_structure_score"]
```

```python
df3.head()
```

```python
def compute_keyword_score(row):
    student_answer = str(row["Student Answer"]).strip().lower()
    keywords_raw = str(row["Keywords"]).strip().lower()

    if student_answer == "" or keywords_raw == "":
        return 0.0

    student_tokens = set(student_answer.split())
    keywords = [kw.strip() for kw in keywords_raw.split(",") if kw.strip()]

    if len(keywords) == 0:
        return 0.0

    matched = 0
    for kw in keywords:
        if set(kw.split()).issubset(student_tokens):
            matched += 1

    max_keyword_marks = float(row["max_marks"]) * 0.10
    keyword_score = (matched / len(keywords)) * max_keyword_marks

    
    return round(keyword_score, 2)
```

```python
df3["keyword_score"] = df3.apply(compute_keyword_score, axis = 1)
df3["keyword_score"]
```

```python
def compute_grammar_score(row):
    try:
        text = str(row["Student Answer"]).strip()

        if text == "":
            return 0.0

        max_grammar_marks = float(row["max_marks"]) * 0.10
        score_ratio = 0

        if text[0].isupper():
            score_ratio += 0.25

        if text.endswith((".", "!", "?")):
            score_ratio += 0.25


        if not re.search(r"[!@#$%^&*]{2,}", text):
            score_ratio += 0.25

        if len(text.split()) >= 5:
            score_ratio += 0.25

        return round(max_grammar_marks * score_ratio, 2)

    except Exception:
        return 0.0
```

```python
df3["grammar_score"] = df3.apply(compute_grammar_score, axis=1)
df3["grammar_score"]
```

```python
def compute_coherence_score(row):
    try:
        text = str(row["Student Answer"]).strip()

        if text == "":
            return 0.0

        max_coherence_marks = float(row["max_marks"]) * 0.05
        score_ratio = 0.0

        sentences = re.split(r'[.!?]', text)
        sentences = [s.strip() for s in sentences if s.strip()]

        if len(sentences) >= 2:
            score_ratio += 0.25

        avg_len = sum(len(s.split()) for s in sentences) / len(sentences)
        if avg_len >= 5:
            score_ratio += 0.25

        words = text.lower().split()
        unique_ratio = len(set(words)) / max(len(words), 1)
        if unique_ratio >= 0.6:
            score_ratio += 0.25

        connectors = ["because", "therefore", "however", "thus", "also", "and", "so"]
        if any(conn in text.lower() for conn in connectors):
            score_ratio += 0.25

        return round(max_coherence_marks * score_ratio, 2)

    except Exception:
        return 0.0
```

```python
df3["coherence_score"] = df3.apply(compute_coherence_score, axis=1)
df3["coherence_score"]
```
